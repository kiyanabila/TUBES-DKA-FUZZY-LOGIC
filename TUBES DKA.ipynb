


## Membuat Fungsi Keanggotaan

```python
## Membuat fungsi segitiga
def triangular(x, a, b, c):

    if x <= a or x >= c:
        return 0

    elif x == b:
        return 1

    elif x < b:
        return (x - a) / (b - a)

    else:
        return (c - x) / (c - b)
        
hasil = triangular(7, 5, 7, 8)

print(hasil)
```

## Membership Function Sleep Duration

```python
def sleep_pendek(x):
    return triangular(x, 5.8, 6.0, 6.5)

def sleep_normal(x):
    return triangular(x, 6.0, 7.0, 8.0)

def sleep_panjang(x):
    return triangular(x, 7.0, 8.0, 8.5)

sleep = 6.5

print("Pendek :", sleep_pendek(sleep))
print("Normal :", sleep_normal(sleep))
print("Panjang :", sleep_panjang(sleep))
```

## Membership Function Stress Level


```python
def stress_rendah(x):
    return triangular(x, 3, 4, 5)

def stress_sedang(x):
    return triangular(x, 4, 5.5, 7)

def stress_tinggi(x):
    return triangular(x, 6, 7, 8)

stress = 7

print("Rendah :", stress_rendah(stress))
print("Sedang :", stress_sedang(stress))
print("Tinggi :", stress_tinggi(stress))
```
